## Notebook that generates 3D graphs that show where the detected colours are in LUV space
Data is currently a direct copy and paste from other files, rather than being nicely generated. To re run with new data, please replace `points`


```python
## import cells
import plotly.express as px
import pandas as pd
import puck.code_modules.colour_conversion.colour_conversion as conv
```

`points` is currently grabbed directly from ``. To update please rerun and grab data from that.

```python
points = [[129, 124, 124],
          [ 92, 130, 122],
          [107, 132, 123],
          [ 98, 130, 119],
          [143, 123, 124],
          [143, 123, 124],
          [113, 138, 119],
          [ 99, 132, 118],
          [104, 135, 121],
          [140, 121, 114],
          [124, 124, 118],
          [129, 137, 126],
          [ 99, 129, 124],
          [106, 128, 123],
          [ 87, 126, 124],
          [ 87, 126, 124],
          [ 95, 123, 122]]
```

Convert the list of points into arrrays made of x coordinates, y coordinates, and z coordinates.


x,y,z all correspond to l,u,v values but for simplicity sake whilst graphing we're keeping them in the standard 3D coordinate system.


An array with the length of the points of "col" is created called "classification". This allows us to colour code all the data points as one colour.

```python
x = [p[0] for p in points]
y = [p[1] for p in points]
z = [p[2] for p in points]
classification = ["col"]*len(points)
```

b_rgb and f_rgb are the raw rgb values of the two colours that are regularly confused for each other. They are accordingly converted to LUV space so they can be graphed with the rest of the detected data.
Use the rgb that b and f are detected at from the calibration sheets, not their "true" encoded values when printed, this will show you where they are relative to the detected data and how that data is getting classified relative to them.

Uncomment the print statement if you want to confirm you are getting the proper luv values.

```python
b_rgb= [134, 60, 123]
b_luv =conv.rgb_to_luv(b_rgb)

f_rgb = [153, 25, 84]
f_luv =conv.rgb_to_luv(f_rgb)

# print(b_luv, f_luv)
```

A previous version of this code used a dataframe as the data source for the figure, it currently isn't in use, but is commented out in case you wish to return to that data structure.

```python
# Create a dataframe where x,y,z, and the classification arrays make up the columns, labelled "x", "y", "z", and "class" accordingly.
# Then add in b's x, y, z values, as well as a classification of "b" at the end of the data frame. Repeat this for f.
# This creates a dataframe called 'data' which can be used to populate the fig.

# # data = pd.DataFrame({"x":x,"y":y,"z":z, "class":classification})
# # data.loc[len(data)]=({"x":b_luv[0], "y":b_luv[1], "z": b_luv[2], "class": "b"})
# # data.loc[len(data)]=({"x":f_luv[0], "y":f_luv[1], "z": f_luv[2], "class": "f"})
```

Create a colour list that assigns a uniform colour for all of the data points, then append the rgb colours of b and f. 

I use the "true" encoded colours so I can see those dots more clearly but you can use the rgb values that are found with the calibration sheet if you prefer.

```python
colour_list = ['rgb(200, 200, 200)'] * len(x)
colour_list.append('rgb(255, 0, 255)')
colour_list.append('rgb(219, 79, 137)')
```

Add in b's x, y, z values, to each coordinate array. Repeat this for f.

```python
x.append(b_luv[0])
x.append(f_luv[0])

y.append(b_luv[1])
y.append(f_luv[1])

z.append(b_luv[2])
z.append(f_luv[2])
```

Create a px.scatter_3d object called fig, where x,y,z, and color are set to their respective arrays. 

To make it clear which colours are associated with which dots add in `color_discrete_sequence=['rgb(200, 200, 200)','rgb(255,0,255)','rgb(219, 79, 137)',` 

where the first rbg is the data, second is b and third is f's rgb colour codes according to how you put it in the colour_list.

I have also updated the marker size, adjust this according to your screen and however small or big you need the markers to be.

Update the layout of the figure, in this case I've adjusted the background to try to make it clearer where the data points are and the lines which make up that grid.

You can colour code each plane of the 3d graph, but I've kept them uniform to reduce visual noise. I do not think highlighting the planes help much in this use case, but it's there if you want it.

```python
fig = px.scatter_3d(x=x, y=y, z=z, color = colour_list, color_discrete_sequence=['rgb(200, 200, 200)','rgb(255,0,255)','rgb(219, 79, 137)',])
fig.update_traces(marker_size = 13)

fig.update_layout(scene = dict(
                    xaxis = dict(
                         gridcolor="white",
                         showbackground=True,
                         gridwidth=10,
                         zerolinecolor="white",),
                    yaxis = dict(
                        gridcolor="white",
                        showbackground=True,
                        gridwidth=10,
                        zerolinecolor="white"),
                    zaxis = dict(
                        gridcolor="white",
                        gridwidth=10,
                        showbackground=True,
                        zerolinecolor="white",),),
                  )
```
