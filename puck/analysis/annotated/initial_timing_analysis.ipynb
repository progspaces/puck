### Notebook to check the timings of Binary, Adaptive Mean, Adaptive Gaussian, and OTSU

This is a mini test to see if the timings are feasible for all the methods. 

If they are not, we should not test them in a wide scale hyper parameter search.

```python
# Import cell
import json
import  puck.experiments.best_timing.binary_otsu_adaptive_timing as boat
```

output_file is the path of the .json file where the output of running boat.main() is saved

adjustment_on adds in the prefix determined by adjustment, which is default, "../", and is off.

cli_printout prints out some metadata to the command line interface, because you're using the notebook to access this, I recommend turnning it off.


```python
output_file = boat.main(adjustment_on=True, cli_printout=False)
```

Check that the `output_file` contains a path string and nothing has been screwed up with the adjustment or lack thereof.

```python
print(output_file)
```

Load in the file data as a dictionary,
Typical key values are:
- `binary_avg` 
- `adaptive_mean_avg`
- `adaptive_gaussian_avg`
- `otsu_avg`

and the same prefixes repeated with `avg` replaced by `median` or `correct`

The `avg` and `median` values are the average or median time it took to run those methods in nanoseconds.

```python
with open(output_file , "r") as json_file:
    file_data = dict(json.loads(json_file.read()))
```

Check that you are getting back a value for the binary method, probably in the 10,000,000 range.

```python
file_data.get("binary_avg")
```

Convert all of these nanosecond values into millisecond values. 

Remember that we need to be in range of about 33 milliseconds per run for the system to run at 30 frames per second.
Any more and it would be unlikely to run in near real time. Any less is fine.

```python
avg_dict_ms = {k: file_data[k]/1_000_000 for k in file_data if k in ["binary_avg", "adaptive_mean_avg", "adaptive_gaussian_avg", "otsu_avg"]}
avg_dict_ms
```

```python
median_dict_ms = {k: file_data[k]/1_000_000 for k in file_data if k in ["binary_median", "adaptive_mean_median", "adaptive_gaussian_median", "otsu_median"]}
median_dict_ms
```

From this we can tell that the adaptive methods both take at least a factor of 10 longer to run, which disqualifies them from being at or near real-time methods. 

We should therefore not run extremely large hyper paremeter tests on these as they are infeasible for the project.

```python
acc_dict = {k: file_data[k]/96 for k in file_data if k in ["binary_correct", "adaptive_mean_correct", "adaptive_gaussian_correct", "otsu_correct"]}
acc_dict
```

From just this prelminary test, it looks like binary is the most promising, method, and given the adaptive methods are under 50% correct and well over time, it seems advisble to not consider them further.

The OTSU being so low in correct results is concerning but perhaps a hyper parameter search will find a more suitable parameterization of the method to return a higher accuracy.
