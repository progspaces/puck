```python
## Imports cell

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statistics as stats
import puck.experiments.best_thresholding.binary_otsu_updated as bou
```

output_file_dict is a dictionary that returns the paths where the outputs of running bou.main() are saved


adjustment_on adds in the prefix determined by adjustment, which is default, "../",


cli_printout prints out some metadata to the command line interface, because you're using the notebook to access this, I recommend turnning it off.


Typical dictionary keys are "bigpicture0", "bigpicture1", "results0", results1", "output timing", where
-  bigpicture# values are the aggregated results of all the possible pipelines using binary or otsu methods in a .csv, indicated by the # 0 for binary, and 1 for otsu.
- results# values are the folder paths where each parameter search is saved as a csv, similarly indicated by the # 0 for binary, and 1 for otsu.
- output timing's value is a .txt file that tells you the time it took to run all of this



A progress bar will show up beneath the cell that is running this, you can use this to approximate how long the cell will take to run. The more hyperparameters tested, the more time it will likely take.

```python
output_file_dict = bou.main(
    adjustment_on = True,
    cli_printout = False
    )
```
